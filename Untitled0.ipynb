```python
from googletrans import Translator

# Create the translator object
translator = Translator()

# Recursive function to translate a list of sentences
def translate_sentences(sentences, dest='ja'):
    if not sentences:  # base case: no sentences left
        return []
    # Translate the first sentence
    translated = translator.translate(sentences[0], dest=dest).text
    # Recursively translate the rest
    return [translated] + translate_sentences(sentences[1:], dest=dest)

# Example sentences to translate
english_sentences = [
    "Hello",
    "How are you?",
    "I love anime",
    "One Piece is amazing",
    "nigga",
]

print("="*50)
print("ENGLISH TO JAPANESE TRANSLATOR")
print("="*50)
print()

# Get the Japanese translations
japanese_translations = translate_sentences(english_sentences, dest='ja')

# Print original and translated sentences
for eng, jap in zip(english_sentences, japanese_translations):
    print(f"English:  {eng}")
    print(f"Japanese: {jap}")
    print("-"*50)

# Now translate from Japanese to English
print()
print("="*50)
print("JAPANESE TO ENGLISH TRANSLATOR")
print("="*50)
print()

# Get the English translations (translating Japanese back to English)
english_translations = translate_sentences(japanese_translations, dest='en')

# Print original Japanese and translated English sentences
for jap, eng in zip(japanese_translations, english_translations):
    print(f"Japanese: {jap}")
    print(f"English:  {eng}")
    print("-"*50)
```
